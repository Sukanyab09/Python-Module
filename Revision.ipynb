Strings- sequential data: you can run loops on strings



```python
greeting = "Welcome to the session. PYTHON Session starts at 12 : 30 PM"
print(greeting)
```

Let's separate the words present in greeting string according to their type(lowercase,uppercase,digits). Save the data in separate lists according to their type and display them at the end


```python
title = [Welcome,Session]
lower_data = [to,the,session]
upper_data = [PYTHON,PM]
digits = [12,30]
```

```python
greeting.split() # separate the data on basis of space
```

```python
# for i in greeting:
#     print(i) # W
#     # e

# create an empty lists to capture title words,lower,upper,digits
title = []
lower_data = []
upper_data = []
digits_data = []
others = []

for i in greeting.split(): 
    # i: Welcome
    # i: to
    # i: the....
    if i.istitle():
        # Welcome.istitle() : True
        # to.istitle() : False
        title.append(i)
    elif i.islower():
        lower_data.append(i)
    elif i.isupper():
        upper_data.append(i)
    elif i.isdigit():
        digits_data.append(i)
    else:
        others.append("Invalid data")
        print("Title",title,",Lower",lower_data,",Upper",upper_data,",Digits",digits_data,",Others",others)
        # f string
        print(f"Title {title},Lower {lower_data},Upper {upper_data},Digits {digits_data},Others {others}")
```
